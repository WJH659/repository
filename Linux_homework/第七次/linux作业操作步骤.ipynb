创建一个docker-work的文件夹
下面创建两个文件
Dockerfile（没有后缀名）
docker-compose.yml

编写Dockerfile

```python
FROM ollama/ollama:latest
ENV DEBIAN_FRONTEND=noninteractive
EXPOSE 11434
CMD ["ollama", "serve"]
```

编写docker-compose.yml

version: "3.8"

services:
  ollama:
    image: docker.m.daocloud.io/ollama/ollama:latest
    container_name: ollama
    command: serve
    ports:
      - "11434:11434"
    volumes:
      - ollama_data:/root/.ollama
    environment:
      - OLLAMA_HOST=0.0.0.0
    restart: unless-stopped

volumes:
  ollama_data:

打开powershell,cd进入刚刚的文件夹docker-work
运行docker-compose up -d
运行docker ps
完成后看到：
CONTAINER ID   IMAGE     COMMAND        STATUS       PORTS
xxxxxxxx      ollama    "/bin/ollama"   Up X sec   0.0.0.0:11434->11434/tcp

接下来运行docker exec -it ollama ollama pull qwen3:0.6b

测试api
创建文件test.py


import requests

response = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": "qwen3:0.6b",
        "prompt": "hello",
        "stream": False
    }
)

print(response.json()["response"])

python test.py运行

打开 Cherry Studio​ 软件。
点击左下角或右上角的 设置图标（齿轮形状）。
在设置菜单中，找到并点击 “模型服务”​ 选项卡。
在服务商列表中找到并点击 “Ollama”。
进行详细配置：
启用状态：打开右侧的开关。
API 密钥：Ollama 默认不需要 API 密钥，将其留空即可。
API 地址：填写 http://localhost:11434/（因为我们在本地电脑部署，且刚才做了端口映射）。
点击下方 “添加”​ 或 “+”。
在弹出的输入框中，模型标识​ 填写：qwen3:0.6b。
点击保存。
